```python

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# ---- Config (edit as you like) ----
FONT_SIZE = 14
CORNER_FIGSIZE = (9, 9)
POINT_ALPHA = 0.7
POINT_SIZE = 20  # for seaborn (points^2)
PLOT_TITLE = "Corner & 3D visualization"

REQUIRED_COLS = ["Delta Lc,nm", "Contour length,nm", "Force, pN"]

# For saving outputs
output_dir = "plots_out"
os.makedirs(output_dir, exist_ok=True)

print("Output directory:", os.path.abspath(output_dir))
```

```python

# >>>> EDIT THIS to point to your Excel file <<<<
excel_path = "your_data.xlsx"  # e.g., "data/my_measurements.xlsx"
excel_path
```

```python

def load_excel_required_columns(file_path, required_cols, rename_map=None):
    # Load Excel and return a clean DataFrame with the required columns only.
    df = pd.read_excel(file_path)
    if rename_map:
        df = df.rename(columns=rename_map)
    missing = [c for c in required_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns: {missing}. Found columns: {list(df.columns)}")
    sub = df[required_cols].apply(pd.to_numeric, errors="coerce")
    sub = sub.replace([np.inf, -np.inf], np.nan).dropna(how="any")
    return sub

df = load_excel_required_columns(excel_path, REQUIRED_COLS, rename_map=rename_map)
df.head()
```

```python
g = sns.pairplot(
    df,
    corner=True,
    plot_kws={"alpha": POINT_ALPHA, "s": POINT_SIZE, "linewidth": 0},
    diag_kind="hist",
    diag_kws={"bins": 40, "alpha": 0.9}
)
g.fig.suptitle(f"{PLOT_TITLE} — Corner plot", y=1.02, fontsize=FONT_SIZE + 2)
g.fig.set_size_inches(*CORNER_FIGSIZE)

corner_png = os.path.join(output_dir, "corner_plot.png")
corner_svg = os.path.join(output_dir, "corner_plot.svg")
g.fig.savefig(corner_png, dpi=300, bbox_inches="tight")
g.fig.savefig(corner_svg, format="svg", bbox_inches="tight")
plt.close(g.fig)

print("Saved:")
print("-", os.path.abspath(corner_png))
print("-", os.path.abspath(corner_svg))
```
