```python

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# ---- Config (edit as you like) ----
FONT_SIZE = 14
CORNER_FIGSIZE = (9, 9)
POINT_ALPHA = 0.7
POINT_SIZE = 20  # for seaborn (points^2)
PLOT_TITLE = "3D visualization"

REQUIRED_COLS = ["Delta Lc,nm", "Contour length,nm", "Force, pN"]

# For saving outputs
output_dir = "C:/Users/lupe184g/Desktop/Postdoc/01_Projects/OT_data/Fiona/Final_thesis_plots"
os.makedirs(output_dir, exist_ok=True)

print("Output directory:", os.path.abspath(output_dir))
```

```python

# >>>> EDIT THIS to point to your Excel file <<<<
excel_path = "C:/Users/lupe184g/Downloads/total_results_H18_full_sorted&analysed1.xlsx"  # e.g., "data/my_measurements.xlsx"
excel_path
```

```python

def load_excel_required_columns(file_path, required_cols, rename_map=None):
    # Load Excel and return a clean DataFrame with the required columns only.
    df = pd.read_excel(file_path)
    if rename_map:
        df = df.rename(columns=rename_map)
    missing = [c for c in required_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns: {missing}. Found columns: {list(df.columns)}")
    sub = df[required_cols].apply(pd.to_numeric, errors="coerce")
    sub = sub.replace([np.inf, -np.inf], np.nan).dropna(how="any")
    return sub

df = load_excel_required_columns(excel_path, REQUIRED_COLS, rename_map=rename_map)
df.head()
```

```python
x, y, z = REQUIRED_COLS  # ["Delta Lc,nm", "Contour length,nm", "Force, pN"]

fig = px.scatter_3d(
    df,
    x=x, y=y, z=z,
    opacity=0.85,
    title=f"{PLOT_TITLE} — 3D scatter",
    height=700
)
fig.update_traces(marker=dict(size=3))
fig.update_layout(
    scene=dict(
        xaxis_title=x,
        yaxis_title=y,
        zaxis_title=z
    ),
    title_x=0.5
)

# Show inline (Jupyter) and also save an HTML you can open in a browser
fig.show()

html_path = os.path.join(output_dir, "interactive_3d_scatter.html")
fig.write_html(html_path, include_plotlyjs="cdn", full_html=True)

print("Saved HTML:")
print("-", os.path.abspath(html_path))

# Optional static export if kaleido is installed
try:
    png_path = os.path.join(output_dir, "interactive_3d_scatter.png")
    svg_path = os.path.join(output_dir, "interactive_3d_scatter.svg")
    fig.write_image(png_path, scale=2)  # requires kaleido
    fig.write_image(svg_path)           # requires kaleido
    print("Also saved static 3D snapshots:")
    print("-", os.path.abspath(png_path))
    print("-", os.path.abspath(svg_path))
except Exception as e:
    print(f"(Static image export skipped: {e})")
```
