```python
# ---- Config (edit as you like) ----
FONT_SIZE = 14
CORNER_FIGSIZE = (9, 9)
POINT_ALPHA = 0.7
POINT_SIZE = 20  # for seaborn (points^2)
PLOT_TITLE = "Corner plot"

REQUIRED_COLS = ["Delta Lc,nm", "Contour length,nm", "Force,pN", "step number"]

# For saving outputs
output_dir = "C:/Users/lupe184g/Desktop/Postdoc/01_Projects/OT_data/Fiona/Final_thesis_plots"
os.makedirs(output_dir, exist_ok=True)

print("Output directory:", os.path.abspath(output_dir))
```

```python

# >>>> EDIT THIS to point to your Excel file <<<<
excel_path = "C:/Users/lupe184g/Downloads/total_results_H18_full_sorted&analysed1.xlsx"  # e.g., "data/my_measurements.xlsx"
excel_path
```

```python

def load_excel_required_columns(file_path, required_cols):
    # Load Excel and return a clean DataFrame with the required columns only.
    df = pd.read_excel(file_path)
    missing = [c for c in required_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns: {missing}. Found columns: {list(df.columns)}")
    sub = df[required_cols].apply(pd.to_numeric, errors="coerce")
    sub = sub.replace([np.inf, -np.inf], np.nan).dropna(how="any")
    return sub

df = load_excel_required_columns(excel_path, REQUIRED_COLS)
df.head()
```

```python
g = sns.pairplot(
    df,
    corner=True,
    x_vars=["Delta Lc,nm","Contour length,nm","Force,pN"],
    y_vars=["Delta Lc,nm","Contour length,nm","Force,pN","step number"],
    plot_kws={"alpha": POINT_ALPHA, "s": POINT_SIZE, "linewidth": 0},
    diag_kind="hist",
    diag_kws={"bins": 40, "alpha": 0.9}
)
g.fig.suptitle(f"{PLOT_TITLE} — Corner plot", y=1.02, fontsize=FONT_SIZE + 2)
g.fig.set_size_inches(*CORNER_FIGSIZE)

png_path = os.path.join(output_dir, f"{PLOT_TITLE}.png")
svg_path = os.path.join(output_dir, f"{PLOT_TITLE}.svg")
g.fig.savefig(png_path, dpi=300, bbox_inches="tight")
g.fig.savefig(svg_path, format="svg", bbox_inches="tight")
plt.close(g.fig)

print("Saved:")
print("-", os.path.abspath(corner_png))
print("-", os.path.abspath(corner_svg))
```
